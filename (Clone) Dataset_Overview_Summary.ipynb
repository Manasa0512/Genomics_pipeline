# Pipeline Architecture Overview
## Medallion Architecture: Bronze → Silver → Gold

**Architecture Pattern**: Medallion (Multi-hop)  
**Implementation**: Databricks Delta Lake + Unity Catalog  
**Status**: [COMPLETE] & Operational

---

## Why Medallion Architecture?

The medallion architecture organizes data into three progressive layers of quality:

### Benefits
* **Incremental Transformation**: Process data in stages, easier to debug
* **Data Quality Zones**: Clear separation of raw, cleaned, and analytics data
* **Reusability**: Silver layer can feed multiple Gold tables
* **Auditability**: Track data lineage through each layer
* **Performance**: Optimize each layer for its purpose
* **Reproducibility**: Reprocess any layer without affecting others

---

## Key Design Decisions

### Layer Purposes

| Layer | Purpose | Write Mode | When to Use |
|-------|---------|------------|-------------|
| **Bronze** | Preserve raw data exactly as received | Append | Historical audit trail |
| **Silver** | Clean, validated, queryable | Overwrite | Direct table queries |
| **Gold** | Denormalized, aggregated, optimized | Overwrite | Analytics & reporting |

### Integration Strategy

**Challenge**: How to join 3 datasets with different schemas and coordinate systems?

**Solution**: Two-stage join in Gold layer

1. **Stage 1**: VCF ↔ GTF (Range Join)
   * Join condition: `vcf.pos BETWEEN gtf.start AND gtf.end`
   * Challenge: Chromosome naming ("1" vs "chr1")
   * Solution: Normalize VCF chromosomes to match GTF
   * Result: 77.7% variants mapped to genes

2. **Stage 2**: (VCF+GTF) ↔ ClinVar (Position Join)
   * Join condition: `chrom = chromosome AND pos = start_pos`
   * Challenge: 99% missing allele data in ClinVar
   * Solution: Position-based join (not allele-based)
   * Result: 73,319 clinical annotations added

## Stage 1: VCF ↔ GTF Join

### Goal: Map variants to genes ("Which gene does this variant affect?")

### Challenge:
* VCF has single position: `(chrom, pos)`
* GTF has position ranges: `(seqname, start, end)`
* Different chromosome formats: "1" vs "chr1"

### Solution: Range Join with Normalization

#### Step 1: Normalize Chromosome Names
```python
# VCF uses "1", GTF uses "chr1" - need to match!
vcf_normalized = silver_vcf_variants.withColumn(
    "chrom_for_gtf",
    when(col("chrom").startswith("chr"), col("chrom"))
    .otherwise(concat(lit("chr"), col("chrom")))
)
# Result: "1" → "chr1"
```

#### Step 2: Filter GTF to Genes Only
```python
# GTF has 5.9M records (genes, transcripts, exons, etc.)
# We only need GENES for variant mapping
gtf_genes = silver_gene_annotations.filter(
    col("feature") == "gene"
)
# Result: 5,868,517 → 78,691 records (98.7% reduction!)
```

#### Step 3: Range Join (Position BETWEEN Start and End)
```python
# Join condition: variant position falls within gene boundaries
vcf_with_genes = vcf_normalized.join(
    broadcast(gtf_genes),  # Broadcast small table (78K rows)
    (
        # Match chromosomes
        (col("vcf.chrom_for_gtf") == col("gtf.seqname")) &
        # Check if position falls in gene range
        (col("vcf.pos") >= col("gtf.start_pos")) &
        (col("vcf.pos") <= col("gtf.end_pos"))
    ),
    "left"  # Keep ALL variants (even without gene match)
)
```

## Summary: The "Primary Key" Answer

### There is NO single primary key column shared among all three tables.

Instead, we use a **composite key**:

```python
JOIN_KEY = (chromosome, position)
```

**With different join strategies:**

| Join | Columns Used | Join Type | Why |
|------|-------------|-----------|-----|
| **VCF ↔ GTF** | `(chrom, pos)` ↔ `(seqname, start, end)` | Range (BETWEEN) | Genes span ranges |
| **VCF ↔ ClinVar** | `(chrom, pos)` ↔ `(chromosome, start_pos)` | Exact (=) | Both are point data |

**Both joins are LEFT joins** to preserve all variants, even those without gene or clinical annotations.

### Code Example:

```python
# Stage 1: VCF + GTF
vcf_with_genes = vcf_normalized.join(
    broadcast(gtf_genes),
    (
        (col("vcf.chrom_for_gtf") == col("gtf.seqname")) &
        (col("vcf.pos") >= col("gtf.start_pos")) &
        (col("vcf.pos") <= col("gtf.end_pos"))
    ),
    "left"  # Keep all variants
)

# Stage 2: (VCF + GTF) + ClinVar
gold_variant_summary = vcf_with_genes.join(
    clinvar,
    (
        (col("v.chrom") == col("c.chromosome")) &
        (col("v.pos") == col("c.start_pos"))
    ),
    "left"  # Keep all variants
)
```

## Key Takeaway

**Genomic data integration requires understanding biology:**
* Variants are **points** on chromosomes
* Genes are **ranges** on chromosomes  
* Clinical data is **sparse** (disease focus)
* Position-based joins are the only reliable method
* Must handle coordinate system differences
* LEFT joins preserve all variants for complete analysis

This is why **77.7% gene mapping and 0.99% clinical annotation are both successful outcomes** - they reflect biological reality, not data quality issues.

## Bronze Layer Optimizations (5)

### 1. Partitioning by ingestion_date
**What it does**: Groups data by day into separate folders on disk  
**Benefit**: Time-based queries are fast (partition pruning), easy data lifecycle management (delete old dates)  
**Example**: `WHERE ingestion_date = '2026-05-28'` only scans one partition instead of entire table

### 2. Delta Lake Format
**What it does**: Uses Delta Lake instead of plain Parquet  
**Benefit**: ACID transactions, time travel, schema evolution, faster reads/writes, automatic compaction  
**Example**: Can rollback bad ingestion, query historical versions, add columns without breaking pipeline

### 3. Append Mode
**What it does**: Each run adds new records without deleting old ones  
**Benefit**: Preserves historical data, creates complete audit trail, enables time-series analysis  
**Example**: Can track when each batch was ingested and compare data quality over time

### 4. inferSchema=false
**What it does**: Reads all data as strings during ingestion  
**Benefit**: No data loss (preserves leading zeros, special formats), no type conversion errors at ingestion  
**Example**: "00123" stays "00123" instead of becoming 123, "N/A" stays "N/A" instead of null

### 5. mergeSchema=true
**What it does**: Allows adding new columns in future runs  
**Benefit**: Pipeline won't break when schema evolves, flexible for future enhancements  
**Example**: Can add new audit metadata columns without recreating tables

---

## Silver Layer Optimizations (5)

### 6. Partitioning by Chromosome
**What it does**: Groups variants by chromosome (chr1, chr2, etc.) into separate folders  
**Benefit**: Chromosome-specific queries skip irrelevant partitions (partition pruning = 22x faster)  
**Example**: `WHERE chrom = 'chr1'` only reads chr1 partition, not all 22 chromosomes

### 7. Overwrite Mode
**What it does**: Each run replaces entire table with latest clean version  
**Benefit**: No duplicate accumulation, simpler than merge logic, always latest data  
**Example**: Daily runs always produce fresh validated data without merge conflicts

### 8. Type Casting & Validation
**What it does**: Converts strings to proper types (int, double, long) with validation  
**Benefit**: Improves query performance (no runtime casting), catches bad data early, enables numeric operations  
**Example**: Position as LONG enables `WHERE pos BETWEEN 1000 AND 2000`, quality as DOUBLE enables `AVG(quality_score)`

### 9. Deduplication
**What it does**: Removes duplicate records during transformation  
**Benefit**: Reduces storage by 50% (ClinVar: 9M → 4.5M), improves query speed, accurate aggregations  
**Example**: ClinVar has duplicate chromosome annotations; dedup ensures accurate counts

### 10. Feature Filtering
**What it does**: Filters GTF to 'gene' features only before loading (5.8M → 78K)  
**Benefit**: 98.7% reduction, dramatically faster downstream joins, less memory usage  
**Example**: Gold layer joins only need 78K genes, not 5.8M transcript/exon/UTR features

## Gold Layer Optimizations (5)

### 11. Broadcast Join
**What it does**: Small table (GTF genes ~78K) sent to all executors instead of shuffle  
**Benefit**: Eliminates expensive shuffle for large VCF table (6.4M records), massive speedup (10-50x faster)  
**Example**: Gene table broadcast to all workers, each worker joins locally without network I/O

### 12. Z-ORDER Clustering
**What it does**: Co-locates related data on disk by gene_name and clinical_significance  
**Benefit**: Queries filtering/grouping by these columns read fewer files (data locality)  
**Example**: `WHERE gene_name = 'BRCA1'` reads only files with BRCA1 data, not entire table

### 13. LEFT Joins
**What it does**: Preserves all variants even without gene/clinical annotations  
**Benefit**: No data loss, complete analysis possible, accurate coverage metrics (77.7% gene mapping)  
**Example**: Intergenic variants (22.3%) retained for completeness analysis

### 14. Range Join Strategy
**What it does**: Optimized genomic coordinate join (position BETWEEN start AND end)  
**Benefit**: Specialized for biological data patterns, accurate variant-to-gene mapping  
**Example**: Finds which gene contains each variant by checking if variant position falls within gene boundaries

### 15. Regexp Extract for Population Data
**What it does**: Extracts population frequencies directly from INFO field using regex  
**Benefit**: Avoids expensive explode operations on nested structures, 6 columns in one pass  
**Example**: Single regex scan extracts AFR_AF, AMR_AF, EAS_AF, EUR_AF, SAS_AF, AF in parallel

**Question**: What percentage of protein-coding vs lncRNA genes have clinical variants?

**Query**:
```sql
SELECT 
    g.gene_type,
    COUNT(DISTINCT h.gene_name) as genes_with_variants,
    SUM(h.total_variants) as total_variants,
```
